# metadata 저장 

```python
import os
import pandas as pd
from experiments.utils import modify_residues_by_id
from data.utils import read_pkl

# -----------------------------------------------------
# 1. 설정 및 데이터 로드
# -----------------------------------------------------
src_metadata_path = "/home/psh/BioBetter/SEZ6/meta/metadata.csv"
save_metadata_path = "/home/psh/BioBetter/SEZ6/meta/metadata_h3_single.csv"
src_pkl = "/home/psh/BioBetter/SEZ6/meta/GZR01T_A_B_C.pkl"
target_dir = '/home/psh/BioBetter/SEZ6/meta'

# 메타데이터 로드 및 초기화 (열과 첫 번째 행만 유지)
df = pd.read_csv(src_metadata_path)
df = df.iloc[[0]].copy()  # 첫 번째 행만 남기고 나머지는 제거

# Pickle 데이터 로드 (Chain ID 확인용)
# read_pkl 구현에 따라 use_torch=False가 필요할 수 있습니다.
data = read_pkl(src_pkl) 
chain_ids = data['chain_index']

# 변이 생성할 아미노산 목록
ALL_AAs = ['ALA','ARG','ASN','ASP','CYS','GLN','GLU','GLY','HIS','ILE','LEU','LYS','MET','PHE','PRO','SER','THR','TRP','TYR','VAL']

# -----------------------------------------------------
# 2. Mutation 수행 및 경로 수집
# -----------------------------------------------------
save_paths = []

# chain_ids[0]은 첫 번째 residue의 체인 ID (보통 numpy/tensor 스칼라)
target_chain = chain_ids[0] 

print("==== Mutation 생성 시작 ====")
for aa in ALL_AAs:
    # modify_residues_by_id 함수 호출 (단일 포인트 변이)
    # mutations=[(체인ID, 잔기번호, 새로운아미노산)]
    save_path = modify_residues_by_id(
        src_pkl, 
        target_dir, 
        mutations=[(target_chain, 103, aa)]
    )
    save_paths.append(save_path)

# -----------------------------------------------------
# 3. 메타데이터 업데이트
# -----------------------------------------------------
print("\n==== 메타데이터 업데이트 시작 ====")

for path in save_paths:
    # 1. 현재 데이터프레임의 '마지막 행'을 복사
    last_row = df.iloc[[-1]].copy()
    
    # 2. 파일명 파싱 (확장자 .pkl 제거)
    filename = os.path.basename(path)
    pdb_name_new = os.path.splitext(filename)[0] # 예: GZR01T_A_B_C_0_103ALA
    
    # 3. 값 교체
    last_row['pdb_name'] = pdb_name_new
    last_row['processed_path'] = path
    last_row['raw_path'] = path
    
    # 4. 데이터프레임 마지막에 붙여넣기
    df = pd.concat([df, last_row], ignore_index=True)

# -----------------------------------------------------
# 4. 결과 저장
# -----------------------------------------------------
# 원본 파일에 덮어쓰기 (혹은 _updated.csv로 저장하려면 경로 변경)
df.to_csv(save_metadata_path, index=False)

print(f"✅ 모든 작업 완료. 업데이트된 메타데이터 저장됨: {save_metadata_path}")
print(f"총 행 개수: {len(df)}")
```
